```python
data = pd.read_csv('transformed/past_house_results.csv')
data = data[data['year'] == 2012]
data.head(10)
```

```python
data.shape
```

```python
data['district_code'] = data['state_po'] + '-' + data['district'].map(lambda x: f'0{x}' if x < 10 else str(x))
data.head(3)
```

```python
trainset = pd.read_csv('transformed/data_2014.csv')
trainset.head(8)
```

```python
trainset.shape
```

```python
trainset = trainset[['district', 'demo_cluster', 'dem_2p_prev', 'dem_2p_prev2', 'prev_lean', 'prev2_lean']]
data = pd.merge(left=data, right=trainset, left_on='district_code', right_on='district', how='left')
data = data.rename({'district_x': 'seat_number'}, axis=1).drop(['district_y'], axis=1)
data.head()
```

```python
data = data[(data['dem_cand'] != '[]') & (data['rep_cand'] != '[]')]
for party in ['dem', 'rep']:
    data[f'{party}_inc'] = data[f'{party}_inc'].map(lambda x: ast.literal_eval(x))
    data[f'{party}_inc_any'] = data[f'{party}_inc'].map(lambda x: x if not isinstance(x, list) else (True if True in x else False))
    data[f'{party}_inc_dummy'] = data[f'{party}_inc_any'].map(lambda x: 1 if x == True else 0)
data.head()
```

```python
data['dem_cand'] = data['dem_cand'].map(lambda x: dem_cand_stand[x] if x in dem_cand_stand.keys() else x)
data['rep_cand'] = data['rep_cand'].map(lambda x: rep_cand_stand[x] if x in rep_cand_stand.keys() else x)
```

```python
data = data.rename({'district_code': 'district'}, axis=1)
data.head(1)
```

```python
full_trainset = pd.read_csv('transformed/all_2p_house_races_trainset.csv')
full_trainset.head(2)
```

```python
incl_cols = ['year', 'state', 'district', 'dem_cand', 'rep_cand', 'dem', 'rep', 'dem_pct_2p', 'dem_inc_dummy', 'rep_inc_dummy', 'dem_2p_prev', 'dem_2p_prev2', 
             'dem_tot_funds', 'rep_tot_funds', 'dem_funds_2p_pct',
            'rep_funds_2p_pct', 'demo_cluster']
war_data = pd.concat([data[incl_cols], full_trainset[incl_cols]], axis=0)
war_data['is_midterm'] = war_data['year'].map(lambda x: 0 if x % 4 == 0 else 1)

## Utilizing Split Ticket's SHAVE data
## Source: https://split-ticket.org/house-generic-ballot-estimates-2008-2022-shave/
shave = {
    2012: -2.4,
    2014: 5.06,
    2016: 1.63,
    2018: -7.26,
    2020: -2.09,
    2022: 1.59,
    2024: 2.15
    
}
war_data['shave'] = war_data['year'].map(shave)
war_data.head()
```
