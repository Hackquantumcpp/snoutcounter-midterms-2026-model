```python
import pandas as pd
import numpy as np
from keys import GOOG_API_KEY
import gspread
from gspread_dataframe import set_with_dataframe
from google.oauth2.service_account import Credentials
```

```python
# Define the scope
scopes = [
    "https://www.googleapis.com/auth/spreadsheets",
    "https://www.googleapis.com/auth/drive"
]

# Authenticate with the service account
credentials = Credentials.from_service_account_file(
    GOOG_API_KEY,
    scopes=scopes
)

# Authorize gspread
client = gspread.authorize(credentials)
```

```python
## Import dataframes
choropleth_data = pd.read_csv('display_data/choropleth_display_data.csv')
table_data = pd.read_csv('display_data/table_display_data.csv')
summary_stats = pd.read_csv('display_data/summary_stats.csv')
output_over_time = pd.read_csv('model_output/output_over_time.csv')
```

```python
## Some wrangling
for party in ['dem', 'rep']:
    choropleth_data[f'disp_{party}_chance'] = choropleth_data[f'disp_{party}_chance'].astype(str)
```

```python
spreadsheet = client.open("snoutcounter-data")
worksheet = spreadsheet.worksheet("senate-forecast-2026-choropleth-data")
worksheet.clear()

set_with_dataframe(worksheet, choropleth_data, string_escaping='full')
```

```python
worksheet = spreadsheet.worksheet("senate-forecast-2026-simple-table-data")
worksheet.clear()

set_with_dataframe(worksheet, table_data, string_escaping='full')
```

```python
worksheet = spreadsheet.worksheet("senate-forecast-2026-summary-stats")
worksheet.clear()

set_with_dataframe(worksheet, summary_stats, string_escaping='full')
```
