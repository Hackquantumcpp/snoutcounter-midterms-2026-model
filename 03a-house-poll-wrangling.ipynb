```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
polls = pd.read_excel('data/polls/Rawpolls_040626.xlsx')
polls.head()
```

```python
polls['rv'] = polls['comment'].str.contains(r'among registered voters').fillna(False)
polls['all_adu'] = polls['comment'].str.contains(r'among all adults').fillna(False)
polls['v'] = polls['comment'].str.contains(r"among 'voters'").fillna(False)
polls.head(2)
```

```python
polls['sponsor'] = polls['comment'].str.extract(r'for ([A-za-z\.\s\,\-\d\|]+).*')
polls.head()
```

```python
np.unique(polls['sponsor'].astype(str))
```

```python
polls['rv'].value_counts()
```

```python
polls['population'] = polls.apply(lambda x: 'A' if x['all_adu'] == True else ('RV' if x['rv'] == True else 
                                                                              ('V' if x['v'] == True else 'LV')), axis=1)
polls.head()
```
